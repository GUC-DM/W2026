### Looping over dictionaries

````python
months = {
    "January": 1,
    "Feburary": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

# looping over keys
````

````python
for month_number in months.values(): # looping over values
    print(month_number)
````

````python
for month_name, month_number in months.items(): # looping over key and value pairs
    print(month_name, month_number)
````
